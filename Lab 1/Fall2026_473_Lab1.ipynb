## Tensors in PyTorch

In PyTorch, all variables representing information in multi-dimensional space, such as vectors and matrices, are referred to as tensors.

Vectors $v \in \mathbb{R}^d$ is a 1D tensor, and matrices $A \in \mathbb{R}^{m\times n}$ is a 2D tensor.

👉 Why do we care about tensors?
Almost every kind of data in machine learning can be stored in a tensor. For example:
- A **grayscale image** is a 2D tensor (rows × columns).
- A **color image** is a 3D tensor (rows × columns × color channels).
- A **batch of images** is a 4D tensor (batch × rows × columns × channels).

Even neural networks can be represented using tensors!

We’ll play with small tensors here, but keep in mind that the same ideas power large models, too!

### Tensor Dimension Reduction

Suppose we are given a 2D tensor (which is basically a matrix) $A\in\mathbb{R}^{m\times n}$. Reduction operations (e.g. sum, mean, std) computes the statistics along a particular axis of the tensor. Examples


*   sum(A, axis=0) gives a vector in $\mathbb{R}^{m}$ whose $i$th entry is the sum $\sum_{j=1}^nA_{ij}$.
*   mean(A, axis=1) gives a vector in $\mathbb{R}^{n}$ whose $j$th entry is the mean $\frac{1}{m}\sum_{i=1}^mA_{ij}$.


```python
A = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32) # 2 x 3 matrix with m = 3, n = 2
print('A: ', A)

A_sum0 = torch.sum(A, axis=0) # m array
print('sum(A, axis=0): ', A_sum0)

A_mean1 = torch.mean(A, axis=1) # n array
print('mean(A, axis=1):', A_mean1)
```

### Matrix transpose

For a matrix $A\in \mathbb{R}^{m\times n}$, its transpose $A^T \in \mathbb{R}^{n\times m}$ is
$$
 [A^T]_{ij} = A_{ji}
$$

If $A = A^T$, then $A$ is symmetric.

```python
A = torch.tensor([[5, 2, 7], [6, 0, 3], [3, 2, 8]])

i, j = (1, 2)

print('[A^T]_{ij}: ', A.T[i,j])
print('A_{ji}: ', A[j,i])
```

## Norms and Inner Products between Vectors


Let's revisit concepts from vector spaces and measure some fundamental properties of a vector. If $x$ is a three dimensional vector defined as:

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)
x
```

The magnitude of $x$ can be represented as the Euclidean norm (or $\ell_2$ norm) of $x$, given by $\|x\|_2$.

The Euclidean norm of $x$ is computed by the squared values at each dimension summed together:
$$
\|x\|_2 = (\sum_{i=1}^d x_i^2)^{\frac{1}{2}}
$$

```python
magX = torch.sqrt(x[0]**2 + x[1]**2 + x[2]**2)
print("Magnitude of x using naive implementation: ", magX)

magX_torch = torch.norm(x)
print('Magnitude of x in PyTorch is:', magX_torch)
```

Measuring this is useful for comparing errors or keeping model weights under control. E.g., to measure the difference betwee a prediction $u$ from the ground-truth $v$, we can compute $\|u - v\|_2$.

There are other types of norms, e.g., the $\ell_1$ norm:
$$
  \|x\|_1 = \sum_{i=1}^d |x_i|
$$

Similarly, we can compare geometric properties of two vectors using these fundamental properties. Imagine a second vector y with the following three dimensional definition:


```python
y = torch.tensor([2, 0, 5], dtype=torch.float32)
```

The magnitude of y is:

```python
magY = torch.sqrt(y[0]**2 + y[1]**2 + y[2]**2)
print("Magnitude of y is: ", magY)
```

We can compute the inner (dot) product between x and y, which will give us a measure of the geometric similarity between the two vectors:

$$
  \langle x, y\rangle = x^\top y = \sum_{i=1}^d x_iy_i
$$

```python
torch.dot(x, y)
```

Using the magnitudes and the dot product, we can actually compute the angle between the two vectors, also defined as the cosine similarity, which measure how closely oriented are two vectors in Euclidean space:

$$
  \langle x, y\rangle = \|x\|_2\|y\|_2 \cos (x, y)\;\; \Rightarrow\;\; \angle (x, y) = \arccos\frac{\langle x, y\rangle}{\|x\|_2 \|y\|_2}
$$

```python
torch.rad2deg(torch.arccos(torch.dot(x, y) / (magX*magY)))
```

Two vectors are orthogonal to each other if the inner product between them is zero:

$$
  x\perp y \Leftrightarrow \langle x, y\rangle = 0
$$

```python
x = torch.tensor([1, 0])
y = torch.tensor([0, 1])
print(torch.dot(x, y))
```

Inner product between $x$ and itself gives the squared Eulidean norm

$$
  \langle x, x \rangle = \sum_{i=1}^d x_i^2 = \|x\|_2^2
$$

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)

print('Inner (x, x): ', torch.dot(x, x))
print('x Norm Squared: ', torch.norm(x) ** 2)
```

## Tensor Multiplication

Vector inner products are tensor multiplications between two 1D tensors. Next we will look at tensor multiplications between higher order tensors.

### Matrix-vector Multiplication

Let $A\in\mathbb{R}^{m\times n}$ and $x\in\mathbb{R}^n$. The matrix-vector product between $A$ and $x$ is given by $Ax \in \mathbb{R}^m$

$$
  [Ax]_i = \sum_{j=1}^n A_{ij}x_j
$$

**Column perspective**: If $A = [a_1, \dots, a_n]$ with $a_j\in \mathbb{R}^m$ for $j\in[n]$, then
$$
  Ax = \sum_{j=1}^n x_j a_j
$$

**Row perspective**: If $v_i$ is the $i$th row of $A$, then
$$
  [Ax]_i = \langle v_i, x\rangle
$$

```python
A = torch.tensor([[1, 2], [3, 4], [5, 6]])
x = torch.tensor([1, 3])

print('A: ', A)
print('x:', x)

Ax = A @ x
print('Ax: ', Ax)
```

```python
a1 = A[:, 0]
a2 = A[:, 1]
print('x1 * a1 + x2 * a2', x[0] * a1 + x[1] * a2)

v1 = A[0, :]
print('<v1, x> (should be [Ax]_1): ', torch.dot(v1, x))
```

### Matrix-matrix Multiplication

For a matrix $A \in \mathbb{R}^{m\times n}$ and a matrix $B \in \mathbb{R}^{n\times p}$, multiplying $A$ with $B$ gives $C = AB \in \mathbb{R}^{m\times p}$

$$
  C_{ij} = \sum_{k=1}^nA_{ik}B_{kj}
$$

**Column-row view**: if the $k$th column of $A$ is $a_i$ and the $k$th row of $B$ is $b_i$, then
$$
  C = \sum_{k=1}^na_kb_k^\top
$$

**Row-column view**: if the $i$th column of $A$ is $u_i$ and the $j$th row of $B$ is $v_j$, then

$$
  C_{ij} = \langle u_i, v_j\rangle
$$


```python
A = torch.tensor([[1, 2], [3, 4], [5, 6]]) # 3x2 matrix
B = torch.tensor([[1, 1, 1], [4, 6, 4]]) # 2x3 matrix

print('A: ', A)
print('B: ', B)

AB = A @ B
print('AB: ', AB)
```

```python
a1, a2 = A[:, 0].reshape((-1, 1)), A[:, 1].reshape((-1, 1)) # Columns of A
b1, b2 = B[0, :].reshape((-1, 1)), B[1, :].reshape((-1, 1)) # Rows of B

print('a_1 b_1^t + a_2 b_2^t: \n', a1 @ b1.T + a2 @ b2.T)

i, j = 1, 2
ui = A[i, :] # ith row of A
vj = B[:, j] # jth column of B

print('<u_i, v_j>: ', torch.dot(ui, vj))
```

Generally, $AB \neq BA$

```python
AB = A @ B
print('AB: ', AB) # 3x3 matrix

BA = B @ A
print('BA: ', BA) # 2x2 matrix
```

## Eigenvalues and Eigenvectors

For a matrix $A\in\mathbb{R}^{n\times n}$, its eigenvector and eigenvalue are pair $(u, \lambda)$ with $u\in\mathbb{R}^n, u \neq 0$ and $\lambda \in\mathbb{R}$ such that
$$
  Au = \lambda u
$$

$A$ should have exactly $n$ eigenvalue-eigenvector pairs, though some eigenvalues are repetitive.

If $A$ is symmetric, then the $n$ eigenvectors of $A$ are orthogonal to each other, and
$$
  A = \sum_{i=1}^n\lambda_iu_iu_i^\top
$$

```python
A = torch.tensor([[2, 0, 3], [1, 4, 6], [3, 7, 10]], dtype=torch.complex128)
L, U = torch.linalg.eig(A)
print("The eigenvectors are: ", U)
print("The eigenvalues are: ", L)
```

## Matrix Inverse

Let $A\in\mathbb{R}^{m\times n}$. Then:


*   $B\in\mathbb{R}^{n\times m}$ is a left-inverse of $A$ if $BA = I$
*   $B\in\mathbb{R}^{n\times m}$ is a right-inverse of $A$ if $AB = I$

Let $A \in \mathbb{R}^{n\times n}$. Then $A$ is invertible if there is a matrix $B$ that is both the left- and the right-inverse of $A$. In this case $B$ is called the inverse of $A$, denoted by $B = A^{-1}$.

## Linear equations in PyTorch

PyTorch contains many useful linear algebra functions for solving systems of equations, computing inverses, etc. Solving linear systems underlies regression and many ML methods.

Below is an example where we use it to solve a system of linear equations $Ax = b$ for the vector $x$.

$$
\begin{bmatrix}1 & 2 & 4 \\ -2 & 3 & 1 \\ -4 & 1 & 2\end{bmatrix}\begin{bmatrix}x_1 \\x_2 \\x_3 \end{bmatrix} = \begin{bmatrix}18 \\ 5 \\ 0\end{bmatrix}
$$

To solve $Ax = b$, we notice that if $A$ is invertible,
$$
  x = A^{-1}Ax = A^{-1}b
$$

We can thus invert the matrix $A$ and then multiply with $b$ to obtain $x$.

```python
# Set up the equations.
import torch

A = torch.FloatTensor([[1, 2, 4], [-2, 3, 1], [-4, 1, 2]])
b = torch.FloatTensor([18, 5, 0])

A_inv = torch.linalg.inv(A)

x = A_inv @ b
print('x = ', x)

# Compute the Euclidean norm of the residual.  It should be 0.
norm_r = torch.linalg.norm(A @ x - b)
print('norm_r = ', norm_r)
```

```python
# Solve the system.
x = torch.linalg.solve(A, b)
print('x = ', x)

# Verify the solution by checking the residual is 0.
print('r = ', A @ x - b)
```

### (Optional) Inverse of Non-degenerate Symmetric Matrices

## Using Matplotlib with PyTorch

In machine learning programs typically there are many variables and information that can be difficult to track by relying on simple logging or printing. We also introduce Matplotlib, the plotting library of Python to aid visualize information related to our machine learning systems in the future.

In Python, we import the following functionalities to set-up any plotting function:

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
```

Let's use matplotlib with torch tensors to plot non-linear functions.

```python
f = lambda x: torch.sin(x)
x = torch.arange(0,10,0.1)

print('x: ', x)
print('f(x): ', f(x))

# Now plot the results.
plt.figure(1)
plt.plot(x, f(x), 'b-', label = r'sine')
plt.xlabel(r'$x$')
plt.ylabel(r'$\sin(x)$')
plt.title(r'The sine function')
plt.legend();
```

We can also plot surfaces representing multi-dimensional information.

```python
x, y = torch.meshgrid(torch.arange(-7, 7, 0.5), torch.arange(-7, 7, 0.5))
z = torch.pow(x, 2) + torch.pow(y, 2)

fig, ax = plt.subplots(subplot_kw={"projection": "3d"}, figsize=(8,8))
surf = ax.plot_surface(x, y, z, cmap = cm.coolwarm)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.title(r'Sphere Function: $z = x^2 + y^2$');
```
